```python
import sherpa_utils

DATASET_DIR = "../../../../../datasets"
```

```python
mock_context_df = sherpa_utils.sherpa_preprocess(f"{DATASET_DIR}/mockDataset.csv")
mock_cpu_df = sherpa_utils.sherpa_get_feature(mock_context_df, "i-0ec321a1c8ed4915c", "CPUUtilization")

display(mock_cpu_df.head())
print(mock_cpu_df.dtypes)
```

```python
import requests

payload = {
    "series": [
        {
            "resource_id": "i-0ec321a1c8ed4915c",
            "resource_type": "ec2",
            "metric_type": "CPUUtilization",
            "timestamps": mock_cpu_df["timestamp"].dt.strftime("%Y-%m-%dT%H:%M:%S.%f").tolist(),
            "values": mock_cpu_df["target"].tolist()
        }
    ],
    "prediction_length": 288
}
```

```python
response = requests.post("http://localhost:5000/forecast", json=payload)

print(response.json())
```

```python
import requests

def make_sample_forecast_request() -> dict:
    return {
        "forecast_horizon": 30,
        "timestamps": mock_cpu_df["timestamp"].dt.strftime("%Y-%m-%dT%H:%M:%S.%f").tolist(),
        "values": mock_cpu_df["target"].tolist(),
        "model": "chronos_univariate",
    }
response = requests.post("http://localhost:5000/forecast-chronos", json=make_sample_forecast_request())
print(response.json())
```
